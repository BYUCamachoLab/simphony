# Discrete Filter

Our discrete filter implementation is similar to matlab, with one alteration: we allow for an optional "delay compensation" parameter. When set to 0, the output of the OpticalDiscreteFilter will mimic traditional implementations. 

When the delay compensation parameter is set to a positive integer, k, the output will be advanced (if possible) by k time steps and an appropriate phase corrected will be applied to complex signals not at the baseband frequency. This allows for more accurate sample mode simulations.

In block mode simulations delay compensation should be set to 0. An error will be thrown otherwise.

```python
from simphony.signals.block_mode import BlockModeOpticalSignal
from functools import partial

def envelope_fn_full(wavelength, t):
    t = jnp.atleast_1d(t)
    T = len(t) # Number of time steps 
    L = 1 # Number of wavelengths
    M = 1 # Number of modes
    A_t = jnp.exp(-0.5*((t-4e-12)/(5e-13))**2)
    amplitude = A_t.reshape((T, L, M))
    
    signal = BlockModeOpticalSignal(amplitude=amplitude, wavelength=jnp.atleast_1d(wavelength))
    
    return signal

wavelength = 1.54e-6
envelope_fn = partial(envelope_fn_full, wavelength)

t = jnp.linspace(0.0, 10e-12, 1000)
envelope = envelope_fn(t)


plt.plot(t, envelope.amplitude[:, 0, 0])
```

```python
import numpy as np
from scipy.signal import dlti, dimpulse

# --- Define a simple discrete-time system ---
A = np.array([[0.5+0.2j, 0], [0, 0.2+0.1j]], dtype=complex)
B = np.array([[1.0], [0.5]], dtype=complex)
C = np.array([[1.0, 1.0]], dtype=complex)
D = np.array([[0.0]], dtype=complex)
dt = 1.0  # sample time

system = dlti(A, B, C, D, dt=dt)

# --- Analytic impulse response ---
N = 10  # number of samples
h_analytic = []

for n in range(N):
    if n == 0:
        h_analytic.append(D.flatten())
    else:
        h_analytic.append((C @ np.linalg.matrix_power(A, n-1) @ B).flatten())

h_analytic = np.array(h_analytic)

# --- Impulse response using dimpulse ---
t, h_dimpulse = dimpulse(system, x0=np.zeros((A.shape[0],), dtype=complex), n=N)
h_dimpulse = np.squeeze(h_dimpulse[0])  # only one input/output

# --- Compare ---
print("Analytic h[n]:")
print(h_analytic.flatten())
print("\ndimpulse h[n]:")
print(h_dimpulse)

# Optional: check if they are close
print("\nDifference:")
print(h_analytic.flatten() - h_dimpulse)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Parameters
# -----------------------------
N = 20              # total number of samples
k = 5               # original delta delay
advance = 1         # number of samples to advance
complex_factor = 1 + 1j  # multiply delta by a complex number

# -----------------------------
# Discrete-time impulse responses
# -----------------------------
n = np.arange(N)
h = np.zeros(N, dtype=complex)
h[k] = complex_factor          # h[n] = delta[n-k] * complex_factor

# Advanced impulse response: h_shifted[n] = h[n+advance]
h_shifted = np.zeros_like(h)
if k - advance >= 0:
    h_shifted[k-advance] = complex_factor
# If advance moves it before n=0, it just disappears (causal system)

# -----------------------------
# Unit step function
# -----------------------------
u = np.ones(N, dtype=complex)

# -----------------------------
# Convolutions
# -----------------------------
conv_h = np.convolve(h, u)[:N]          # truncate to original length for simplicity
conv_h_shifted = np.convolve(h_shifted, u)[:N]

# -----------------------------
# Plot results
# -----------------------------
plt.figure(figsize=(12,5))

# Magnitude
plt.subplot(1,2,1)
markers, stemline, baseline = plt.stem(n, np.abs(conv_h), linefmt='b', markerfmt='bo', basefmt='k', label='h * u')
markers.set_alpha(0.5)

markers, stemline, baseline = plt.stem(n, np.abs(conv_h_shifted), linefmt='r', markerfmt='ro', basefmt='k', label='h_shifted * u')
markers.set_alpha(0.5)
plt.title("Magnitude of convolution with unit step")
plt.xlabel("n")
plt.ylabel("|(h*u)[n]|")
plt.legend()

# Phase
plt.subplot(1,2,2)
plt.stem(n, np.angle(conv_h), linefmt='b', markerfmt='bo', basefmt='k', label='h * u')
plt.stem(n, np.angle(conv_h_shifted), linefmt='r', markerfmt='ro', basefmt='k', label='h_shifted * u')
plt.title("Phase of convolution with unit step")
plt.xlabel("n")
plt.ylabel("Phase [rad]")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
def compute_group_delay(wavelengths, s_params):
    # Unwrap phase
    phase = np.unwrap(np.angle(s_params))  # radians

    # Derivative of phase w.r.t wavelength
    dphi_dlambda = np.gradient(phase, wavelengths)  # radians per meter

    # Compute group delay
    c = 299792458  # speed of light in m/s
    group_delay = -dphi_dlambda * (wavelengths**2) / (2 * np.pi * c)  # in seconds

    return group_delay  # same length as input

S = siepic.waveguide(wl=np.linspace(1.5, 1.6, 1000), length=100)[('o0', 'o1')]

tau1 = compute_group_delay(wl, S)[500]
tau2 = 4e-12
plt.plot(t, np.abs(laser_output.amplitude[:, 0, 0]))
plt.plot(t, np.abs(wg_output.amplitude[:, 0, 0]))
# plt.axvline(tau1+tau2+.5*0*1e-12)
# plt.axvline(tau1+tau2+.5*1*1e-12)
# plt.axvline(tau1+tau2+.5*2*1e-12)
# plt.axvline(tau1+tau2+.5*3*1e-12)
# plt.axvline(tau1+tau2+.5*4*1e-12)
```

```python
import numpy as np
import jax.numpy as jnp
import matplotlib.pyplot as plt
from scipy.signal import bessel, freqz, tf2zpk

# === 1. Bessel Filter Design ===
order = 5
cutoff = 0.02  # Normalized frequency (0.5 = Nyquist)
b, a = bessel(order, cutoff, btype='low', analog=False, norm='phase')

# === 2. Group Delay Calculation ===
w, h = freqz(b, a, worN=2048)
freq = w / jnp.pi  # Normalized frequency [0, 1]
phase = jnp.unwrap(jnp.angle(h))
group_delay_bessel = -jnp.gradient(phase) / jnp.gradient(w)

# === 3. Design a Simple All-Pass Filter for Compensation ===
# First-order all-pass: H(z) = (a + z^-1) / (1 + a z^-1)
# We'll pick 'a' close to 1 to give more delay at higher frequencies

a1 = 0.01
b_ap = jnp.array([a1, 1.0])
a_ap = jnp.array([1.0, a1])

# Frequency response of the all-pass
_, h_ap = freqz(b_ap, a_ap, worN=2048)
phase_ap = jnp.unwrap(jnp.angle(h_ap))
group_delay_ap = -jnp.gradient(phase_ap) / jnp.gradient(w)

# === 4. Cascade Bessel + All-Pass ===
from scipy.signal import convolve

b_total = convolve(b, b_ap)
a_total = convolve(a, a_ap)

_, h_total = freqz(b_total, a_total, worN=2048)
phase_total = jnp.unwrap(jnp.angle(h_total))
group_delay_total = -jnp.gradient(phase_total) / jnp.gradient(w)

# === 5. Plot Group Delays ===
plt.figure(figsize=(10, 6))
plt.plot(freq, group_delay_bessel, label='Bessel Filter')
plt.plot(freq, group_delay_ap, label='All-Pass Compensation')
plt.plot(freq, group_delay_total, label='Cascaded (Flattened)')
plt.xlim(0, 0.05)
plt.xlabel('Normalized Frequency (π radians/sample)')
plt.ylabel('Group Delay (samples)')
plt.title('Group Delay Compensation using All-Pass Filter')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np

def smooth_discontinuity(x, f, x0, width):
    # Create a smooth blending weight
    w = 1 / (1 + np.exp(-(x - x0) / width))
    # Assume f is defined on both sides already
    return (1 - w) * f['left'](x) + w * f['right'](x)

# Example usage
f_left  = lambda x: np.zeros_like(x)      # before discontinuity
f_right = lambda x: np.ones_like(x)       # after discontinuity
x = np.linspace(-2, 2, 500)
y_smooth = smooth_discontinuity(x, {'left': f_left, 'right': f_right}, 0, 0.05)

# plt.plot(x, y)
plt.plot(x, y_smooth)
plt.show()
```

```python
import numpy as np

def cubic_hermite_patch(x, xa, ya, ma, xb, yb, mb):
    # Normalized parameter t in [0,1]
    t = (x - xa) / (xb - xa)
    h00 = (1 + 2*t) * (1 - t)**2
    h10 = t * (1 - t)**2
    h01 = t**2 * (3 - 2*t)
    h11 = t**2 * (t - 1)
    return h00*ya + h10*(xb - xa)*ma + h01*yb + h11*(xb - xa)*mb

# Example data: cusp at x=5
x = np.linspace(0, 10, 200)
y = np.abs(x - 5)  # artificial cusp

# Smoothing interval
xa, xb = 4, 6
ya = y[np.argmin(np.abs(x - xa))]
yb = y[np.argmin(np.abs(x - xb))]
ma = (y[np.argmin(np.abs(x - (xa+0.01)))] - ya) / 0.01
mb = (yb - y[np.argmin(np.abs(x - (xb-0.01)))]) / 0.01

# Build smoothed function
y_smooth = y.copy()
mask = (x >= xa) & (x <= xb)
y_smooth[mask] = cubic_hermite_patch(x[mask], xa, ya, ma, xb, yb, mb)

plt.plot(y)
plt.plot(y_smooth)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
N = 256              # number of samples
f_width = 20         # width of the rectangular spectrum
delta_index = 102+ N // 2 # position of Kronecker delta (center)

# Frequency domain: rectangular spectrum
freq = np.fft.fftfreq(N, d=1.0)
rect_spec = np.where(np.abs(freq) < f_width / N, 1.0, 0.0)

# Inverse Fourier transform -> time-domain sinc-like function
time_func = np.fft.ifft(np.fft.ifftshift(rect_spec))

# Kronecker delta
kronecker_delta = np.zeros(N)
kronecker_delta[delta_index] = 1.0

# Convolution (should reproduce time_func)
conv_result = np.convolve(time_func, kronecker_delta, mode='same')

# Plot results
plt.figure(figsize=(10,4))
plt.plot(np.real(time_func), label="Inverse FT of Rect (sinc)")
plt.plot(np.real(conv_result), '--', label="Convolution with delta")
plt.legend()
plt.title("Convolution of sinc with Kronecker delta")
plt.show()
```
