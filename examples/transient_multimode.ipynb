```python
import numpy as np
from simphony.libraries.siepic import y_branch
from functools import partial
import matplotlib.pyplot as plt
from simphony.utils import create_multimode_sax_model


y_branch_models = {
    "te": partial(y_branch, pol="te"),
    "tm": partial(y_branch, pol="tm")
}

wl = np.linspace(1.5, 1.6, 1000)
y_branch_sdict = create_multimode_sax_model(y_branch_models)(wl=wl)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

ax1.plot(wl, np.abs(y_branch_sdict[("port_1@te", "port_2@te")]), label="te")
ax1.plot(wl, np.abs(y_branch_sdict[("port_1@tm", "port_2@tm")]), label="tm")
ax1.legend()

ax2.plot(wl, np.unwrap(np.angle(y_branch_sdict[("port_1@te", "port_2@te")])), label="te")
ax2.plot(wl, np.unwrap(np.angle(y_branch_sdict[("port_1@tm", "port_2@tm")])), label="tm")
ax2.legend()
```
