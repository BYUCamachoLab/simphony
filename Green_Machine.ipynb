# Green Machine Reservoir Computer
This notebook reproduces a Green Machine photonic reservoir computer and documents the full pipeline from signal generation through post-processing.

## Reservoir Computing Primer
An overview of photonic reservoir computing can be found in [Tait et al. 2017](https://www.degruyterbrill.com/document/doi/10.1515/nanoph-2016-0132/html?lang=en&srsltid=AfmBOoo3gODf92DXW8qSAsPjCnuZ9eP5P9Ci6JZY1k9pWJz1nVDwFelT).
Reservoir computers expand the input into a higher-dimensional feature space so that a **linear** readout can solve nonlinear tasks.

For illustration, suppose the target relation is $\mathbf{y}(t) = \mathbf{u}(t)^2$.
If the reservoir exposes the feature $\mathbf{x}(t) = \mathbf{u}(t)^2$, a linear readout $\sum_i w_i x_i(t)$ suffices to approximate $\mathbf{y}(t)$.
The photonic circuit therefore acts as a deterministic projection
$$
\mathbf{x}(t) = f_{\text{res}}(\mathbf{u}(t)),
$$
followed by a linear readout
$$
\hat{y}(t) = \mathbf{w}^{\mathrm{H}} \mathbf{x}(t) + b,
$$
with trainable complex weights $\mathbf{w}$ and bias $b$.
The physical mixing provided by the Green Machine builds this rich feature space.

## Green Machine Architecture
We create the NGRC-inspired circuit from [Ooi et al. 2024](https://www.nature.com/articles/s41467-024-55172-3).
Multiple delay lines feed a 7-channel Green Machine so that each symbol experiences a unique mix of temporally shifted copies.
The optical readout measures
$$
\hat{y}_{\mathrm{PD}}(t) = \left| \sum_{i=1}^{N} w_i x_i(t) \right|,
$$
whose magnitude-squared expands to self-terms and cross-terms, generating expressive reservoir states.


 
In the simulations below we inject Lorenz and NARMA-7 time series into phase modulators, propagate through measured S-parameters, and record the mixed outputs for later linear regression.

## Signal Generation Workflow
The next cell synthesizes the two driving datasets used throughout the notebook:
- a chaotic Lorenz trajectory that is mapped onto an optical phase drive, and
- a stochastic NARMA-7 sequence that serves as a benchmarking signal for the reservoir.

The helper functions perform the following steps:
1. Integrate the continuous-time Lorenz system with SciPy to obtain $x(t), y(t), z(t)$.
2. Normalize and mix the Lorenz coordinates so they fit within the modulator phase range.
3. Generate a NARMA-7 input/output pair using the classic discrete recurrence with a configurable burn-in period.
4. Shape both discrete-time sequences with a root-raised-cosine pulse (``rrcosfilter``) to emulate realistic symbol waveforms at 512 samples per symbol.
These shaped streams ultimately modulate the phase of the write arm in the circuit and the bias/reference arm feeding the delay ladder.

```python
N = 10000  # length of the symbol stream we ultimately feed into the reservoir
def minmax(u):
    return (u - u.min())/(u.max() - u.min())

sigma, rho, beta = 10.0, 28.0, 8/3  # canonical Lorenz parameters

def lorenz(t, xyz):
    x, y, z = xyz
    return [sigma*(y - x), x*(rho - z) - y, x*y - beta*z]

# --- Integrate the Lorenz attractor ------------------------------------------------
N_coarse = N  # number of coarse integration steps
dt = 0.05  # coarse sampling period for the continuous-time system
T = dt * (N_coarse - 1)  # total simulated time span

t_fine = np.arange(0.0, T + 1e-12, dt)

sol_fine = solve_ivp(
    lorenz, (0.0, T), (0., 1., 1.05),
    method="RK45", max_step=dt, t_eval=t_fine
)
x_f, y_f, z_f = sol_fine.y

# Rescale Lorenz coordinates into [0,1] and map onto phase range
signal  = minmax(x_f)
signal2 = minmax(y_f)
y       = minmax(z_f)
combined_sig = signal + signal2
combined_sig = minmax(combined_sig)*jnp.pi  # bias around pi/4

# --- Generate the discrete NARMA-7 benchmark ---------------------------------------
def narma7(N, seed=0, u_low=0.0, u_high=0.5, burn_in=1000):
    rng = np.random.default_rng(seed)
    U = rng.uniform(u_low, u_high, size=N + burn_in + 10)  # random drive
    Y = np.zeros_like(U)
    for t in range(6, N + burn_in):
        s = np.sum(Y[t-6:t+1])  # rolling history term
        Y[t+1] = 0.3 * Y[t] + 0.05 * Y[t] * s + 1.5 * U[t-6] * U[t] + 0.1
    u = U[burn_in+1:burn_in+1+N]  # discard transient
    y = Y[burn_in+1:burn_in+1+N]
    return u, y

u, y_narma = narma7(N, seed=42)  # training pair (drive, target)

# --- Pulse shaping to produce physical waveforms -----------------------------------
sps = 512  # samples per symbol (matches later sampling rate)
beta = 0.35  # roll-off factor
span = 4  # filter spans 4 symbols

_, h = rrcosfilter(span*sps+1, beta, 1.0, sps)  # root-raised-cosine taps

def generate_binary_phases(key, N):
    bits = jax.random.randint(key, (N,), 0, 2)
    return bits * jnp.pi, bits

def rrc_shape(symbols, h, sps):
    x = upfirdn(h.astype(np.float32), np.asarray(symbols, np.float32), up=sps)
    d = (len(h) - 1) // 2  # group delay of the RRC filter
    return jnp.asarray(x[d:d + len(symbols)*sps], dtype=jnp.float32)

def zoh_shape(symbols, sps):
    return jnp.repeat(jnp.asarray(symbols, jnp.float32), sps)

phase_wave1 = rrc_shape((np.pi - minmax(u)*jnp.pi/2), h, sps)  # NARMA drive wrapped into phase
phase_wave2 = rrc_shape(combined_sig, h, sps)  # Lorenz-based phase modulation
```

## Circuit Assembly
This section builds the photonic reservoir netlist in three stages:
1. **Custom components** – lightweight sample-mode wrappers for a phase modulator and streaming multi-line laser source. These expose the ports and stateful behaviour expected by `SampleModeSimulation`.
2. **Green Machine model** – loads measured S-parameters (`gm_real_sdict_1p50_1p60um_1000pts.npz`) and presents them as a SAX-compatible component so the simulator can apply a unitary mixing transform.
3. **Netlist construction** – instantiates lasers, Y-branches, delay waveguides, and the Green Machine, then connects them into the feed-forward delay ladder. Additional loops at the end of the cell replicate the variable-length delay taps described in the NGRC architecture.

Pay close attention to the `settings` dictionary: it captures per-instance parameters (comb wavelengths, delay lengths, modulator drive) that will be handed to the simulator immediately before `run()`.

```python
class OpticalModulatorTime(SampleModeComponent):
    optical_ports = ["o0", "o1"]

    def __init__(self, *, length=1.0, operating_wl=1.55e-6, effective_index=2.4,
                 group_index=4.2, mod_signal=0.0):
        self.length = float(length)
        self.operating_wl = float(operating_wl)
        self.effective_index = float(effective_index)
        self.group_index = float(group_index)
        self.mod_signal = jnp.asarray(mod_signal)  # cached phase drive

    def sample_mode_initial_state(self, simulation_parameters):
        return jnp.int32(0)  # track current sample index through the waveform

    def sample_mode_step(self, inputs, state, simulation_parameters):
        a0 = inputs["o0"].amplitude  # forward field (wavelengths × modes)
        a1 = inputs["o1"].amplitude  # backward field (unused, but keeps port balanced)
        lam = inputs["o0"].wavelength

        if a0.ndim == 1: a0 = a0[:, None]  # force (L,M) shape
        if a1.ndim == 1: a1 = a1[:, None]
        dtype = a0.dtype

        phase_t = self.mod_signal[state]  # select phase drive at current time step

        coeff = jnp.exp(1j * phase_t).astype(dtype)  # phase-only modulation
        out0 = SampleModeOpticalSignal(jnp.zeros_like(a1), lam)  # terminated backward port
        out1 = SampleModeOpticalSignal(a0 * coeff, lam)  # forward field acquires phase

        return {"o0": out0, "o1": out1}, state + 1

def GM(wl: Union[float, Array] = 1.55):
    def load_sdict_npz(fname):
        d = np.load(fname)
        wls = d["wavelengths"]
        S = {}
        for k in d.files:
            if k == "wavelengths": continue
            _, to_p, from_p = k.split("_", 2)
            S[(to_p, from_p)] = np.squeeze(d[k])
        return wls, S
    w_i, Sp = load_sdict_npz("gm_real_sdict_1p50_1p60um_1000pts.npz")
    return Sp  # expose measured interferometer scattering matrix

class StreamingCombSourceTime(SampleModeComponent):
    optical_ports = ["o0"]

    def __init__(self, base_wavelengths, scale=1.0):
        bw = jnp.asarray(base_wavelengths, jnp.float32).reshape(-1)
        self.base  = bw
        self.M     = int(bw.shape[0])
        self.scale = jnp.float32(scale)

    def sample_mode_initial_state(self, simulation_parameters):
        phi = jnp.zeros((self.M,), jnp.float32)  # phases per comb line
        t_idx = jnp.int32(0)  # streaming counter
        return (phi, t_idx)

    @partial(jax.jit, static_argnums=(0,))
    def sample_mode_step(self, inputs, state, simulation_parameters):
        phi, t_idx = state
        A = (self.scale * jnp.exp(1j * phi).astype(jnp.complex64))[:, None]  # per-line amplitude
        wls = self.base.astype(jnp.float32)
        out = {"o0": SampleModeOpticalSignal(amplitude=A, wavelength=wls)}
        return out, (phi, t_idx + jnp.int32(1))  # advance streaming index

# -----------------------------------------------------------------------------------
# Netlist and settings describing the NGRC-style delay ladder feeding the Green Machine
# -----------------------------------------------------------------------------------
netlist = {
    "instances": {
        "laser1":"laser1",
        "laser2":"laser2",
        "gm":"GM",
        "wgmzi1":"waveguide",
        "wgmzi2":"waveguide",
        "pm1":"phase_mod",
        "y1":"y_branch",
        "y2":"y_branch",
        "y3":"y_branch",
        "y4":"y_branch",
        "y5":"y_branch",
        "y6":"y_branch",
        "y7":"y_branch",
        "y8":"y_branch",
        "y9":"y_branch",
        "y10":"y_branch",
        "y_combined":"y_branch",
        "wgd1":"wg_delay",
        "wgd2":"wg_delay",
        "wgd3":"wg_delay",
        "wgd4":"wg_delay",
        "wgd5":"wg_delay",
        "wgd6":"wg_delay",
        "wgd7":"wg_delay",
        "wgd8":"wg_delay",
    },
    "connections": {
        "laser1,o0": "y2,port_1",  # CW comb feeds splitter tree
        "laser2,o0":"wgd1,o0",  # second comb drives primary delay path

        "y2,port_2": "wgmzi1,o0",  # launch into Mach-Zehnder arm
        "y2,port_3": "wgmzi2,o0",
        "wgmzi1,o1":"pm1,o0",  # modulator applied to one arm
        "pm1,o1":"y3,port_2",
        "wgmzi2,o1":"y3,port_3",

        "y3,port_1":"y4,port_1",  # build binary splitter tree
        "y4,port_2":"y5,port_1",
        "y4,port_3":"y6,port_1",
        "y5,port_2":"y7,port_1",
        "y5,port_3":"y8,port_1",
        "y6,port_2":"y9,port_1",
        "y6,port_3":"y10,port_1",

        "y7,port_2":"wgd2,o0",  # launch into staggered delays
        "y7,port_3":"wgd3,o0",
        "y8,port_2":"wgd4,o0",
        "y8,port_3":"wgd5,o0",
        "y9,port_2":"wgd6,o0",
        "y9,port_3":"wgd7,o0",
        "y10,port_2":"wgd8,o0",

        "wgd1,o1":"gm,in0",  # feed delays into Green Machine inputs
        "wgd2,o1":"gm,in1",
        # additional taps appended programmatically below
    },
    "ports": {
        "out0": "gm,out0",
        "out4": "gm,out4",
        "out1": "gm,out1",
        "out5": "gm,out5",
        "out2": "gm,out2",
        "out6": "gm,out6",
        "out3": "gm,out3",
        "out7": "gm,out7",
    }
}

models = {
    "waveguide": siepic.waveguide,
    "laser1": StreamingCombSourceTime,
    "laser2": StreamingCombSourceTime,
    "laser3": StreamingCombSourceTime,  # spare channel if needed later
    "y_branch": siepic.y_branch,
    "phase_mod": OpticalModulatorTime,
    "wg_delay":  siepic.waveguide,
    "GM": GM,
}

dt = 1e-14  # sample period driving the simulator
delay = 366  # base unit delay index (maps to dt via waveguide model)
t = jnp.arange(0, len(phase_wave1))*dt
print(len(t))

settings = {
    "laser1": {
        "base_wavelengths": jnp.array([1.52e-6,1.53e-6,1.54e-6,1.55e-6,1.56e-6], jnp.float32),
        "scale":jnp.sqrt(7.0),  # relative power launched into splitter tree
    },
    "laser2": {
        "base_wavelengths": jnp.array([1.52e-6,1.53e-6,1.54e-6,1.55e-6,1.56e-6], jnp.float32),
        "scale":jnp.sqrt(0.0),
    },
    "wgd1":{
        "length":100
    },
    "wgd2":{
        "length":100
    },
    "wgd3":{
        "length":100
    },
    "wgd4":{
        "length":100
    },
    "wgd5":{
        "length":100
    },
    "wgd6":{
        "length":100
    },
    "wgd7":{
        "length":100
    },
    "wgd8":{
        "length":100 
    },
    "pm1":{
        "mod_signal": phase_wave2,  # Lorenz-derived phase drive
    },
    "gm":{
        "max_order":150  # limit state-space order inside the S-parameter wrapper
    },
}

# Extend delay lattice to replicate NGRC incremental delays
for i in range(6):
    for j in range(i+1):
        settings[f"wg_delay{i}{j}"] = {"length":delay, "max_order": 80}  # tunable per-tap length
        netlist["instances"][f"wg_delay{i}{j}"] = "waveguide"
        netlist["connections"][f"wg_delay{i}{j},o1"] = f"wg_delay{i}{j+1},o0"  # chain tap to next stage
    netlist["connections"][f"wg_delay{i}{j},o1"] = f"gm,in{i+2}"  # terminate final tap at GM input
    netlist["connections"][f"wg_delay{i}0,o0"] = f"wgd{i+3},o1"  # connect source waveguide into ladder

circuit = Circuit(netlist=netlist, models=models)  # validate netlist & expand defaults
sim = SampleModeSimulation(circuit = circuit)  # inserts terminations and JIT plumbing
simulation_parameters = SampleModeSimulationParameters(
    optical_baseband_wavelengths=jnp.array([1.52e-6,1.53e-6,1.54e-6,1.55e-6,1.56e-6], jnp.float32),
    num_time_steps=int(len(t)), sampling_period=dt)

results = sim.run(simulation_parameters=simulation_parameters, settings=settings)  # launch time-domain run
jax.block_until_ready(results)  # make sure computation completes before downstream use
```

## Temporal Eye Analysis
The helper routines below align the sampled waveforms, sweep sampling phase to maximize eye opening, and extract symbol-rate decision samples for downstream learning.

```python
def build_eye_and_best_phase(y, dt, sps, k_ui=2):
    """
    y   : 1D waveform
    dt  : sample period (1e-14 in your case)
    sps : samples per symbol 
    k_ui: window width in unit intervals (2 is standard)
    Returns: (t_ui, M, best_phase, sample_time_offset)
      - t_ui: time axis in UI
      - M:    matrix of shape (num_traces, k_ui*sps) for plotting
      - best_phase: integer [0..sps-1]
      - sample_time_offset: best_phase * dt
    """
    # Sweep phase to maximize "eye opening" at mid-UI:
    # we use variance at the center sample (bigger variance -> crossed transitions align).
    mid = (k_ui * sps) // 2
    best_phase, best_metric = 0, -np.inf

    for p in range(sps):
        seg = y[p:]
        L = (len(seg) // (k_ui * sps)) * (k_ui * sps)
        if L == 0:
            break
        M = seg[:L].reshape(-1, k_ui * sps)
        # metric: variance at center + slope around center (optional combo)
        metric = np.var(M[:, mid])
        if metric > best_metric:
            best_metric = metric
            best_phase = p

    # Build the final eye at that phase
    seg = y[best_phase:]
    L = (len(seg) // (k_ui * sps)) * (k_ui * sps)
    M = seg[:L].reshape(-1, k_ui * sps)
    t_ui = np.linspace(-k_ui/2, +k_ui/2, k_ui * sps, endpoint=False)
    return t_ui, M, best_phase, best_phase * dt


def extract_symbol_samples(y, dt, sps, best_phase, k_ui=2):
    """
    Return the samples taken at the middle of each UI after aligning by best_phase.
    """
    seg = y[best_phase:]
    step = sps  # 1 sample per symbol at decision time
    # center of the 2-UI window:
    decision_index = (k_ui * sps) // 2
    # Align windows then take the center sample of each window
    L = (len(seg) // (k_ui * sps)) * (k_ui * sps)
    M = seg[:L].reshape(-1, k_ui * sps)
    samples = M[:, decision_index]
    # Useful for plotting sampling points on original time axis:
    sample_times = (best_phase + decision_index + np.arange(len(samples)) * (k_ui * sps)) * dt
    return samples, sample_times

# Pick which port & wavelength you want to analyze:
port_to_view = "out0"       # e.g., an MMI output port
wl_index     = 0          # choose the comb line index you care about
y = jnp.abs(results[port_to_view]["output"].amplitude[:, wl_index])**2

sps = 512     # your shaping setting
dt  = 1e-14    # your sim sample period
t_ui, M, best_phase, phase_time = build_eye_and_best_phase(y, dt, sps, k_ui=2)

print(f"Best sampling phase = {best_phase} samples  (~{phase_time*1e12:.2f} ps) from your trace start")

plt.figure(figsize=(6,4.2))
plt.plot(t_ui, M.T, lw=0.5, alpha=0.25)
plt.axvline(0.0, color='k', lw=1.0, alpha=0.5)
plt.title("Eye Diagram")
plt.xlabel("Time [UI]"); plt.ylabel("Amplitude")
plt.tight_layout(); plt.show()


sym_samples, sample_times = extract_symbol_samples(y, dt, sps, best_phase, k_ui=2)
print(f"Collected {len(sym_samples)} decision samples.")
```
