```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# parameters
sigma, rho, beta = 10.0, 28.0, 8/3
N = 10000
dt = 0.05

def lorenz(t, xyz):
    x, y, z = xyz
    return [sigma*(y - x), x*(rho - z) - y, x*y - beta*z]

T = dt * (N - 1)
t_eval = np.arange(0.0, T + 1e-12, dt)

sol = solve_ivp(
    lorenz, (0.0, T), (0., 1., 1.05),
    method="RK45", max_step=dt, t_eval=t_eval
)
x_f, y_f, z_f = sol.y

def minmax(u):
    return (u - u.min())/(u.max() - u.min())

z_lorenz = minmax(z_f)

# drop first 1000, then take next 2000
segment = z_lorenz[1000:3000]

plt.figure(figsize=(10,4))
plt.plot(segment, linewidth=0.8)
plt.title("z_lorenz (normalized), samples 1000–3000")
plt.xlabel("Sample index (offset by 1000)")
plt.ylabel("Normalized z")
plt.tight_layout()
```
